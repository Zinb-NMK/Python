Palindrome List

```python
class Solution:
    #
    def is_palindrome(self, nums):
        left = 0
        right = len(nums) - 1

        while left < right:
            if nums[left] != nums[right]:
                return False
            left += 1
            right -= 1
        return True
try:
    nums = list(map(int, input("Enter a list of nums: ").split()))
    obj = Solution()
    print(obj.is_palindrome(nums))
except ValueError:
    print("enter only integers separated by space")
```

```python
class Solution:

    def is_palindrome(self, num):

        if num < 0:
            return False

        original = num
        reverse = 0
        while num>0:
            digit = num%10
            reverse = reverse * 10 + digit
            num = num // 10
        return original == reverse
try:
    num = int(input("Enter number: "))
    obj = Solution()
    print(obj.is_palindrome(num))
except ValueError:
    print("Enter only integers ")




```

Reverse a List

```python
from array import array
class Solution:
    def reverse_arr(self, nums):
        left = 0
        right = len(nums)-1

        while left < right:
            nums[left], nums[right] = nums[right], nums[left]
            left += 1
            right -= 1
        return nums
try:
    nums = array('i', map(int, input("Enter elements: ").split()))
    obj = Solution()
    print(obj.reverse_arr(nums))

except ValueError:
    print("enter only integers separated by space")
```

Rotate List - Left

```python
class Solution:
    def left_rotate(self, nums, k):

        def reverse(nums, left, right):
            while left < right:
                nums[left], nums[right] = nums[right], nums[left]
                left += 1
                right -= 1
        left = 0
        n = len(nums)

        if k == 0:
            return nums

        k = k % n

        reverse(nums, left, k - 1)
        reverse(nums, k, n - 1)
        reverse(nums, left, n - 1)

        return nums

try:
    nums = list(map(int, input("Enter elements: ").split()))
    k = int(input("Enter k value: "))
    obj = Solution()
    print(obj.left_rotate(nums,k))
except ValueError:
    print("enter only integers separated by space")


```

Plus One in the Number

```python
class Solution:
    def plus_one(self, nums):

        n = len(nums)

        for num in nums:
            if num < 0 or num > 9:
                return "no negative nums: "

        for i in range(n - 1, -1, -1):
            if nums[i] < 9:
                nums[i] += 1
                return nums
            elif nums[i] == 9:
                nums[i] = 0


        return [1] + nums

try:
    nums = list(map(int, input("Enter elements: ").split()))
    obj = Solution()
    print(obj.plus_one(nums))
except ValueError:
    print("enter only integers separated by space")
```

Missing Number

```python
class Solution:
    def missing_num(self, nums):

        n = len(nums)
        actual_sum = n * (n + 1)//2
        current_sum = 0

        for num in nums:
            current_sum += num

        return actual_sum - current_sum
try:
    nums = list(map(int, input("Enter elements: ").split()))
    obj = Solution()
    print(obj.missing_num(nums))
except ValueError:
    print("enter only integers separated by space")
```

Is Array Sorted?

```python
class Solution:
    def is_sorted(self, nums):
        n = len(nums)

        for i in range(1, n):
            if nums[i] < nums[i - 1]:
                return False
        return True

try:
    nums = list(map(int, input("Enter elements: ").split()))
    obj = Solution()
    print(obj.is_sorted(nums))
except ValueError:
    print("enter only integers separated by space")
```

Move Zeroes

```python
class Solution:
    def move_zeros(self,nums):

        n = len(nums)
        k = 0
        for i in range(n):
            if nums[i] != 0:
                nums[k], nums[i] = nums[i], nums[k]
                k+=1
        return nums

nums = list(map(int, input("Enter elements: ").split()))
obj = Solution()
print(obj.move_zeros(nums))


```

Intersection of two Lists

```python
class Solution:
    def intersection_lst(self, num1, num2):
        set1 = set(num1)
        re = set()
        for num in num2:
            if num in set1:
                re.add(num)
        return list(re)
num1 = list(map(int, input("Enter elements: ").split()))
num2 = list(map(int, input("Enter elements: ").split()))
obj = Solution()
print(obj.intersection_lst(num1, num2))
```

Max Consecutive Ones

```python
class Solution:
    def max_consecutive_ones(self, nums):
        n = len(nums)
        max_count = 0
        count = 0

        for i in range(n):
            if nums[i] == 1:
                count += 1
            elif nums[i] == 0:
                count = 0

            if max_count < count:
                max_count = count
        return max_count
try:
    nums = list(map(int, input("Enter list of nums: ").split()))
    obj = Solution()
    print(obj.max_consecutive_ones(nums))
except ValueError:
    print("enter only 0's and 1's separated by space")
```

Maximum Subarray Sum

```python
class Solution:

    def max_sub_array(self, nums):

        max_sum = float('-inf')
        current_sum = 0

        if not nums:
            return 0

        for num in nums:
            current_sum += num

            max_sum = max(max_sum, current_sum)

            if current_sum < 0:
                current_sum = 0

        return max_sum

nums = list(map(int, input("Enter elements: ").split()))
obj = Solution()
print(obj.max_sub_array(nums))
```

Given a number (e.g., 2500):

-> find sum of odd digits

-> find sum of even digits

```python
class solution:
    def so_su(self,n):
        odd_sum = 0
        even_sum = 0
        original = n
        while n > 0:
            dig = n % 10

            if dig % 2 == 0:
                even_sum += dig
            else:
                odd_sum += dig

            n = n // 10

        return f"Given number is {original}\nOdd sum: {odd_sum} \nEven sum: {even_sum}"

n = int(input("Enter a number: "))
obj = solution()
print(obj.so_su(n))
```

Given a paragraph:


-> count total number of words

-> Sort words based on length (min -> max)

-> print odd-length words separately

-> print even-length words separately


```python

# input
para = input("Enter a para: ")

# Split para into words and make a list
words = para.split(" ")

# Total word Count
total_count = len(words)
print(f"Total number of words: {total_count}")

# Sort words based on length
sorted_word = sorted(words, key=len)

print(f"Sorted words Based on len: {sorted_word}")
# odd & even len words.
odd_len_ele = []
even_len_ele = []

for word in words:
    if len(word) % 2 ==0:
        even_len_ele.append(word)
    else:
        odd_len_ele.append(word)




print(f"Even Length Words: {even_len_ele} \nOdd Length Words: {odd_len_ele}")


```
