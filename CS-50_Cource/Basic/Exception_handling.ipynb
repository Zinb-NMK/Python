```python
try:
    a = int(input("Enter a num: "))
except ValueError:
    print("Only integers are accepted")
else:
    print(f"a is {a}")
```

```python
def main():
    x = get_int()
    print(f"x is {x}")

def get_int():
    while True:
        try:
            x = int(input("Enter x value: "))
        except ValueError:
            print("Only integers are accepted. ")
        else:
            break
    return x

main()
```
