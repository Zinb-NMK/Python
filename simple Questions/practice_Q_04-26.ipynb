How to print a calandar for the given month and year in python.

```python
import calendar

def calander_from_Year_Month(y,m):
    """This function finds the calander for the given months and Year"""
    print(calendar.month(y,m))
y = int(input("Enter Year: "))
m = int(input("Enter Months: "))
calander_from_Year_Month(y,m)
```

How to Calculate number of dayss between two dates in python

```python
from datetime import date

def diff_bt_days(s=date(2004,3,5),e=date(2026,1,4)):
    """This function Calculate number of dayss between two dates"""
    delta = e - s
    return delta.days
print(diff_bt_days())
```

Find Volume and surface area of sphere
4*pi*r2
4/3*pi*r3

```python
def vol_sf_of_sphere(r):
    """This function finds the Volume and Surface are of Sphere"""
    sa = 4*3.14*r**2
    vol = 4*3.14*(r**2)*r
    return f"Surface area: {sa}, Volume: {vol}"

r = float(input("Enter raduse of Sphere: "))
print(vol_sf_of_sphere(r))
```

```python
def hallow_square(n):
    """This function prints hollow square"""
    for i in range(n): # Row
        for j in range(n): # Columne
            if i==0 or i==n+1 or j==0 or j==n-1:
                print("* ", end=" ")
            else:
                print(" ", end=" ")
        print()
int1 = int(input())
hallow_square(int1)
```
