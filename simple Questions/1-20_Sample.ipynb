Code Execution Time

```python
import time
def n_my(n):
    start = time.time()
    s = 0
    for i in range(1,n+1):
        s +=i
    end = time.time()
    return s, end - start
n = int(input())
print(n_my(n))
```

```python
def linear_search(n, lst):
    #
    leng = len(lst)
    for index in range(leng):
        if lst[index] == n:
            return index

    return -1
n = int(input("Enter target Element: "))
lst = list(map(int, input("Enter List Elements: ").split()))
print(linear_search(n, lst))

```
