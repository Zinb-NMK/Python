```python
def findmin(nums):
    #
    left = 0
    right = len(nums) - 1

    while left < right:
        mid = (left + right) // 2

        if nums[mid] > nums[right]:
            left = mid + 1
        else:
            right = mid
    return nums[left]

#
nums = list(map(int, input("Enter the array: ").split()))
print(findmin(nums))

```

```python
def min_in_sorted_array(nums):
    #
    start = 0
    end = len(nums) - 1

    while(start < end):
        mid = (start + end) // 2

        if nums[mid] > nums[end]:
            start = mid + 1
        else:
            end = mid
    return nums[start]
nums = list(map(int, input("Enter the array: ").split()))
print(min_in_sorted_array(nums))
```
