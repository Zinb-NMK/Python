```python
def findmin(nums):
    #
    left = 0
    right = len(nums) - 1

    while left < right:
        mid = (left + right) // 2

        if nums[mid] > nums[right]:
            left = mid + 1
        else:
            right = mid
    return nums[left]

#
nums = list(map(int, input("Enter the array: ").split()))
print(findmin(nums))

```

```python
def ser_in_rotated_array(nums,target):
    #
    left = 0
    right = len(nums) - 1

    while left <= right:
        mid = (left + right) // 2

        if nums[mid] == target:
            return mid

        else:

            if nums[left] <= nums[mid]:

                if nums[left] <= target < nums[mid]:
                    right = mid  - 1
                else:
                    left = mid + 1
            else:
                if nums[mid] < target <= nums[right]:
                    left = mid + 1
                else:
                    right = mid - 1

    return -1
nums = list(map(int, input("Enter the array elements: ").split()))
target = int(input("Enter the target value: "))
print(ser_in_rotated_array(nums,target))



```
