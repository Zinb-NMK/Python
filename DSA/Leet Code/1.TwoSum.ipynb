'''Given an array of integers nums and an integer target, return indices of the
two numbers such that they add up to target.

You may assume that each input would have exactly one solution,
and you may not use the same element twice.You can return the answer in any order.'''

```python
# Brute Force Solution.

""" Time Complexity: o(n^2) Because we are using 2 loops.
    Space Complexcity: o(1) Because we are not using any extra space. We are just returning the index of the elements. """

def TwoSum(nums, target):
    #
    n = len(nums)
    for i in range(n):
        for j in range(i+1, n):
            if (nums[i] + nums[j] == target):
                return [i, j]
nums = list(map(int, input("Enter a num: ").split()))
target = int(input("Enter a target: "))
print(TwoSum(nums, target))
```

```python
def Two_Sum(nums, target):
    #

    n = len(nums)
    nums.sort()
    left = 0
    right = n - 1

    while(left < right):
        sum_v = nums[left] + nums[right]

        if(sum_v == target):
            return [nums[left], nums[right]]
        elif(sum_v < target):
            left += 1
        elif(sum_v > target):
            right -= 1
        else:
            return "No Element Add's Up to Target"

nums = list(map(int, input("Enter a numbers: ").split()))
target = int(input("Enter a target: "))
print(Two_Sum(nums, target))
```

```python
def two_sum(nums, target):
    #
    hashmap = {}

    for i, num in enumerate(nums):
        diff = target - num

        if diff in hashmap:
            return [hashmap[diff], i]

        hashmap[num] = i
    return "No Solution"

nums = list(map(int, input("Enter a numbers: ").split()))
target = int(input("Enter a target: "))
print(two_sum(nums, target))
```
