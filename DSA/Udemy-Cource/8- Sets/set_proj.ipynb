Design and develop a Python program that performs the following tasks:

Accept integer values from the user to create two different sets, namely Set A and Set B.

Display the elements of both sets.

Find and display the maximum element present in:

Set A

Set B

Perform the union operation on Set A and Set B and:

Display the union set

Find and display the maximum element in the union set

Perform the intersection operation on Set A and Set B and:

Display the intersection set

Find and display the maximum element in the intersection set

Handle the situation where the intersection set is empty and display an appropriate message.

Ensure the program is user-friendly and follows proper coding standards.

```python
seta = input("Enter Set A Elements: ").split(",")
setb = input("Enter Set B Elements: ").split(",")

setA = set(seta)
setB = set(setb) 
# Elements of SetA
print(f"Elements of SetA are: {setA}\n")
#Elemnts of SetB 
print(f"Elements of SetB are: {setB}\n")

# Max od SetA
print(f"Max of SetA is: {max(setA)}\n")
print(f"Max of SetB is: {max(setB)}\n")

# Union Operatin
un =  setA.union(setB)
print(f"Union of the two sets is: {un}\n")
print(f"Max of union is: {max(un)}\n")

#intersection of both sets is
intr = setA.intersection(setB)


if intr==set():
    print("No Intersection Elements\n")
else:
    print(f"Intersection of two sets is: {intr}\n")
    print(f"Max of intersection is: {max(intr)}\n")




```
