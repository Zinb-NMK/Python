Sets are built-in data types in Python used to store collection of unique items. they are unorderd, meaning that the elements do not follow a specific order, and they do not allow dublicate elements. Sets are useful for membership tests, eleminating dublicate entries, and performing mathematical set operations like union. intersections, difference and symmetric difference.

```python
## Creating a set
my_set = {5.7,1,2,"Manoj",4,6}
print(my_set)
type(my_set)
```

```python
my_set = set([1,1,2,3,4,5,6,77,77,8,99,99])
print(my_set)
```

```python
##Basic set Operations 

##Ading and Removing set elements
my_set.add(8.6)
print(my_set)

## Removing a set element
"""my_set.remove(99)
print(my_set)
"""
```

```python
my_set.discard(132) ## Is an element is not in the set even if we remove that element it won't give an error
print(my_set)
```

```python
## PoP method in set

rem = my_set.pop()
print(rem)
print(my_set)
```

```python
## Clear all elements
my_set.clear()
print(my_set)
```

```python
# Set memberShip test
my_set  = {1,2,3,4,5,6}
print(4 in my_set)
print(10 in my_set)
```

```python
## Mathematical opperations
set1 = {1,2,3,4,5}
set2 = {5,6,7,8,9,4}

## union
union_set = set1.union(set2)
print(union_set)

## intersection
intersection_set = set1.intersection(set2)
print(intersection_set)

set1.intersection_update(set2)
print(set1)
```

```python
## Defereence
s1 = {1,2,3,4,5,6}
s2 = {4,5,6,7,8,9}

print(s1.difference(s2))
print(s2.difference(s1))
```

```python
## Symmetric Difference
s1.symmetric_difference(s2) ## Common elements are removed from both the sets and print unique elements
```

```python
# Subset
s1 = {1,2,3,4,5}
s2 = {1,2,3,4,5,6,7,8,9}
print(s1.issubset(s2))
```

```python
## Unique elements in List
l = [1,1,2,2,3,4,5,5,6,7,8,8,5,7]
print(set(l))
```
