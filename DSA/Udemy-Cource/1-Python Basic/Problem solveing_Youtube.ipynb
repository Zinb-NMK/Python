```python
1. ## How to find average of N numbers in python.

def Average_Of_N_Numbers(num):
    Avg = 0
    Total_Sum = 0 
    for i in range(num):
        numbers = float(input("Enter Numbers: "))
        Total_Sum +=numbers
    Avg = Total_Sum/num
    print(f"Average of Given Numbers is {Avg}")
num = int(input("Enter How Many Numbers ? "))
Average_Of_N_Numbers(num)
```

```python
2. ## How to sum of the First N Positive Integers in Python

def sum_of_n_numbers(n):
    sum = n*(n+1)/2
    print(f"Sum of N Numbers is {sum}")
n = int(input("Enter Value of N ?"))
sum_of_n_numbers(n)

## Or Using For Loop

def sum_of_n_numbers_loop(num):
    sum = 0
    for i in range(n+1):
        sum +=i
    return sum
n = int(input("Enter Number of N ? "))
result = sum_of_n_numbers_loop(n)
print(f"Sum of N numbers is  : {result}")
```

```python
3. ## How to get time of a Python program's Execution 
import time
def Program_Execution_Time(n):
    start_time = time.time()
    sum = 0
    for i in range(n+1):
        sum += i
    end_time = time.time()
    return sum, end_time - start_time
n = int(input("Enter Value of n ? "))
print(Program_Execution_Time(n))
```

```python
10. ## How to find the Area of the Circle
def Area_of_circle(radius):
    area = 3.14 * radius * radius
    return area
radius = float(input("Enter Radius of the circle: "))
print(Area_of_circle(radius))

## OR using math module
import math
def area_of_circle(radius):
    area = math.pi * radius ** 2
    return area
radius = float(input("Enter Radius of the circle: "))
print(area_of_circle(radius))
```

```python
11. ## How to write first_name and last_name in reverse order

def Simple_reverse_name(first_name, last_name):
    return last_name + " " +first_name

first_name = input("Enter First Name: ")
last_name = input("Enter Last Name: ")
print(Simple_reverse_name(first_name, last_name))
      
```

```python
16. ## How to read num and compute n+nn+nnn in Python

def compute_n(n):
    if n>=0:
        n2 = int("%i%i" % (n,n))
        n3 = int("%i%i%i" % (n,n,n))

        return n + n2 + n3
n = int(input("Enter a number: "))
print(compute_n(n))

## OR
def compute_n_loop(n):
    total = 0
    str_n = str(n)
    for i in range(1,4):
        total += int(str_n * i)
    return total
n = int(input("Enter a number: "))
print(compute_n_loop(n))
```
