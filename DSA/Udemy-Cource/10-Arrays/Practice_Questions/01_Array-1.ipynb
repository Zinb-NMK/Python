Description:
Given a list of integers, write a function to find the maximum element in the list.


Input Parameters:

lst (List[int]): A list of integers.

Output:

int: The maximum element in the list.


Example:

Input: lst = [3, 5, 2, 9, 6]
Output: 9
 
Input: lst = [-1, -2, -3, -4]
Output: -1
 
Input: lst = [7]
Output: 7

```python
def max_num(nums):
    #
    max_val = nums[0]

    for num in nums:
        if num > max_val:
            max_val = num
    return max_val

nums = list(map(int, input("Enter the numbers: ").split()))
print("Maximum number in the list is:", max_num(nums))
```

Description:
Given a list of integers, determine if it is a palindrome. A list is considered a palindrome if it reads the same forward and backward.


Input Parameters:

lst (List[int]): A list of integers.

Output:

bool: Return True if the list is a palindrome, otherwise False.

Example:

Input: lst = [7, 8, 9, 8, 7]
Output: True
 
Input: lst = [1, 2, 3, 4, 5]
Output: False
 
Input: lst = [1, 2, 3, 2, 1]
Output: True


```python
def Palindrome(nums):
    #
    left = 0
    right = len(nums)- 1

    while left < right:
        if nums[left] == nums[right]:
            left += 1
            right -= 1
        else:
            return False
    return True

nums = list(map(int, input("Enter a list of nums: ").split()))
print(f"Is the list a palindrome? {Palindrome(nums)}")
```

Description:
Given a list of integers, write a function to reverse the order of elements in the list.


Input Parameters:

lst (List[int]): A list of integers.

Output:

List[int]: The list with elements in reversed order.

Example:

Input: lst = [1, 2, 3, 4, 5]
Output: [5, 4, 3, 2, 1]
 
Input: lst = [10, 20, 30]
Output: [30, 20, 10]
 
Input: lst = [7, 8, 9]
Output: [9, 8, 7]


```python
def reverse_list(nums):
    #

    left = 0
    right = len(nums) - 1

    while left < right:
        nums[left], nums[right] = nums[right], nums[left]
        left += 1
        right -= 1
    
    return nums

nums = list(map(int, input("Enter a list of nums: ").split()))
print(f"Reversed List: {reverse_list(nums)}")
```

```python
# Divide List into 2 halves
# list --> parta + partb [1st_ reverse part, 2nd_ reverse partb, reverse whole list]

def rotate_list(nums, d):
    #
    n = len(nums)
    d = d % n

    def reverse(nums, start, end):
        while start < end:
            nums[start], nums[end] = nums[end], nums[start]

            start += 1
            end -= 1
    
    reverse(nums, 0 , d - 1)
    reverse(nums, d, n - 1)
    reverse(nums, 0 , n - 1)
    
    return nums

nums = list(map(int, input("Enter a list of nums: ").split()))
d = int(input("Enter the number of positions to rotate: "))
print(f"Rotated List: {rotate_list(nums,d)}")
```

Description:
You are given a large integer represented as an integer array digits, where each digits[i] is the i-th digit of the integer. The digits are ordered from most significant to least significant in left-to-right order. The large integer does not contain any leading zeroes.

Write a function to increment the large integer by one and return the resulting array of digits.


Input Parameters:

digits (List[int]): A list of integers where each integer represents a digit of a large number.

Output:

List[int]: The list representing the number after incrementing it by one.


Example:

Input: digits = [1, 2, 3]
Output: [1, 2, 4]
 
Input: digits = [4, 3, 2, 1]
Output: [4, 3, 2, 2]
 
Input: digits = [9, 9, 9]
Output: [1, 0, 0, 0]


```python
def last_plus_one(nums):
    #
    end = len(nums)- 1

    if nums[end] < 0:
        return nums
    
    while end >= 0:
        if nums[end] < 9:
            nums[end] += 1
            return nums
        else:
            nums[end] = 0
            end -= 1
    
    return [1] + nums

nums = list(map(int, input("Enter a list of nums: ").split()))
print(f"List after adding one: {last_plus_one(nums)}")
```

```python
def missing_num(nums):
    #
    n = len(nums)

    Actule_sum = n *(n + 1)// 2

    current_sum = 0
    for num in nums:
        current_sum += num
    
    return Actule_sum - current_sum

nums = list(map(int, input("Enter a list of nums: ").split()))
print(f"Missing number is: {missing_num(nums)}")
```

Description:
Write a function that checks whether the given array is sorted in non-decreasing order. The array is considered sorted if every element is less than or equal to the next element.


Input Parameters:

arr (List[int]): A list of integers.

Output:

bool: True if the array is sorted in non-decreasing order, False otherwise.


Example:

Input: arr = [5, 4, 3, 2, 1]
Output: False
 
Input: arr = [1, 3, 2, 4, 5]
Output: False
 
Input: arr = [1, 2, 3, 4, 5]
Output: True


```python
def sorted_nums(nums):
    #
    n = len(nums)

    for i in range(n - 1):
        if nums[i] > nums[i + 1]:
            return False
    return True
```
