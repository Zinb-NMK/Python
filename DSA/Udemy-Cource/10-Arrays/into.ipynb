Binary Search

```python
def Binary_Search(arr,target):
    #  

    size = len(arr)
    start = 0
    end = size-1

    while( start <= end):
        mid = (start + end )// 2

        if (arr[mid] == target):
            return mid
        
        elif (arr[mid]  > target):
            end = mid - 1
        
        elif (arr[mid] < target):
            start = mid + 1
        
    return -1
arr =list(map(int, input("Enter numbers in assending order: ").split()))
target = int(input("Enter the target number: "))
print(Binary_Search(arr, target))

```
