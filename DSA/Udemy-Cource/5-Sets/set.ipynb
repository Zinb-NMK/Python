Set ignores the dublicate elements.

```python
 
 my_empt = set([1,2,2,3,4,5,6,7,6,8,9])
print(my_empt)
```

Basic Set operations.


```python
# Add elements into Sets
my_empt.add(77)
print(my_empt)
my_empt.add(77)
print(my_empt)
```

```python
# Remove Elements from Sets.

my_empt.remove(77)
print(my_empt)
```

```python
# Remove element if it is present in sets otherwise does Nothing
my_empt.discard(3)
print(my_empt)
```

```python
## pop method == First in First Out
removed_ele = my_empt.pop()
print(removed_ele)
print(my_empt)
```

```python
## Clear All Elements

my_empt.clear()
print(my_empt)
```

Set Membership test ; Checks if the element is present in my set or not

```python
my_set = {1,2,3,4,5}
print(3 in my_set)
print(10 in my_set)
```

Mathematical Operations

```python
st1 = {1,2,3,4,5,6,7}
st2 = {6,7,8,9,10}

#Union 
union_set = st1.union(st2)
print(union_set)

## Intersection
inter_set = st1.intersection(st2)
print(inter_set)

st1.intersection_update(st2)
print(st1)
```

```python
st1 = {1,2,3,4,5,6,7}
st2 = {6,7,8,9,10}

# Difference
print(st1.difference(st2))
```

```python
st2.difference(st1)
```

```python
# Only common Elements are Removed from st1 & st2.
st1.symmetric_difference(st2)
```

```python
## Sets Methods
st1 = {1,2,3,4}
st2 = {3,4}
print(st1.issubset(st2))
print(st1.issuperset(st2))
```
