```python
def type_num(num):
    # Check for Prime
    if num > 1:
        for i in range(2, num):
            if(num%i)==0:
                return f"Num is not a prime"
        else:
            return f"Num is prime"
    else:
        return f"num must be greater then 1 "
num = int(input("Enter a num: "))
Prime = type_num(num)
print(f"Prime: {Prime}\n")

        
```
