Bubble Sort Algorithm

```python
def Bubble_sort(arr):
    #

    n = len(arr) # Arry length
    for passes in range(0, n): #Number of Cycles
        for j in range(0, n-1 - passes): # Number of comparisons
            if(arr[j] > arr[j+1]): # Comparing adjacent elements
                arr[j],arr[j+1] = arr[j+1], arr[j] #swapping if the element is greater than the next element
    return arr
arr = list(map(int, input("Enter a List of Nums: ").split()))
print(Bubble_sort(arr))
```
