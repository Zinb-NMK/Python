# Selection Sorting Algorithem

```python
def select_sort(nums):
    
    n = len(nums) # Findes the length of the list
    
    for i in range(0, n-1): # Loop through the list until the second last element
        min_index = i
        
        for j in range(i+1,n): # Loop through the list starting from the next element of i until the end of the list
            if nums[j] < nums[min_index]: # If the current element is smaller than the minimum element, update the minimum index
                min_index = j
            
        nums[i], nums[min_index] = nums[min_index], nums[i] # Swap the minimum element with the first element of the unsorted list
                
    return nums

nums = list(map(int, input("Enter the numbers: ").split()))
print(select_sort(nums))

    
```
