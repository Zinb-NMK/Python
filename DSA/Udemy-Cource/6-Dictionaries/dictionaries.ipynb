Creating Dictionaries

```python
empt_dict = {}
print(type(empt_dict))
```

```python
empp = dict()
empp
```

```python
stud = {"name":"Manoj", "age":21,"grade":"A+"}
print(stud)

# Single key is always used:
stud = {"name":"Manoj", "age":21,"grade":"A+", "name": "kaju"}
print(stud)
```

```python
#Accesing Elements in Dictionaries
print(stud['grade'])
print(stud['age'])
```

```python
#Accesing using get
print(stud.get('grade'))
print(stud.get('las_name'))
print(stud.get('l_name',"Not Available"))
```

```python
## Modifing Dictionary Elements
## Dictionaries are mutable, So you can add, update or delete elements

print(stud)
```

```python
## Adding Keys to dictionaries
stud["age"] = 20
print(stud)
stud["Gender"] = "Male"
print(stud)
```

```python
## Deleting Keys from Dictionaries
del stud['grade']
print(stud)
```

```python
## Dictionary methods
keys=stud.keys()
print(keys)
values = stud.values()
print(values)

## Key value pairs {List of Tuples}
items = stud.items()
print(items)
```

```python
## Shallow Copy
stud_copy = stud
print(stud)
print(stud_copy)
```

```python
stud["name"] = "kumar"
print(stud)
print(stud_copy)
```

```python
stud_copy1 = stud.copy() ## Shallow copy Provides Different Memory Location. 
print(stud_copy1)
print(stud)
```

```python
stud["name"] = "Kaju"
print(stud_copy1)
print(stud)
```

```python
## Iterating over dictionaries
## We can use Loops to iterate over Dictionaries.

# Iterating over keys.
for key in stud.keys():
    print(key)
```

```python
## Iterating over Values.
for value in stud.values():
    print(value)
```

```python
## Iterate over Key value Pairs
for key, value in stud.items():
    print(f"{key}:{value}")
```

```python
## Nested  Dictionaries
stud = {
    "st1":{'name':"Aksh", 'age':32},
    "st2":{'name':"puja", 'age': 21}
}
print(stud)
```

```python
## Access nested Dictionary Elements
print(stud["st1"]["name"])
print(stud['st2']["age"])
```

```python
stud.items()
```

```python
## Iterating over Nested Dictionaries
for stud_id,stud_info in stud.items():
    print(f"{stud_id}:{stud_info}")
    for key,value in stud_info.items():
        print(f"{key}:{value}")
```

```python
## Dictionaries Comprehension
squa={x:x**2 for x in range(10)}
print(squa)
```

```python
## Conditional Dictionary Comprehension
even = {x:x**2 for x in range(10) if x%2==0}
print(even)
```

```python
# Practical Example

numb = [1,2,2,3,3,3,4,4,5,6,6,7,8,8,9]
freq = {}
for num in numb:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1
print(freq)
```

Merging Two Dictionaries into one

```python
dic1 = {"Name":"Manoj","Age":21,"Gender":"Male"}
dic2 = {"Location":"India", "Phon":1234567}

merge = {**dic1, **dic2}
print(merge)
```
