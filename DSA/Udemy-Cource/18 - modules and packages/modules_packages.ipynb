# In python, modules and packages help organize and reuse code. Here's a comprehensive guide on how to import them.



```python
import numpy as np
np.array([1,2,3,4,5])
```

```python
# It import square root function from math module 
from math import sqrt
print(sqrt(25))
```

```python
# It is used to import all the functions in math module
from math import *

print(sqrt(16))
print(pi)
print(e)
```

```python
from package.maths import addition
addition(10,20)
```

```python
from package import maths
maths.addition(30,30)
```
