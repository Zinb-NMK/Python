## Static Variables and Methods in OOPs

```python
class Atm:

    # Constructor(Special Function
    def __init__(self):
        self.__balance = 0 #Private Variable
        self.pin = ""
        # self.menu()

    # getter method
    def get_balance(self):
        return self.__balance

    # Setter Method
    def set_balance(self, new_balance):
        if type(new_balance) == int:
            self.__balance = new_balance
        else:
            print('Not Allowed')


    def menu(self):
        user_input = int(input("""Enter 1. Create PIN,
        2. Check Balance, 3. Update PIN: ,
        4. Withdraw, 5. Deposit: ,
        any other num to exit():"""))

        if user_input == 1:
            # Create Pin
            self.create_pin()
        elif user_input== 2:
            # Check Balance
            self.check_balance()
        elif user_input == 3:
            # Update Pin
            self.update_pin()
        elif user_input == 4:
            # Withdraw
            self.withdraw()
        elif user_input == 5:
            # Deposit
            self.deposit
        else:
            print("Thank you for using this ATM machine")


    def create_pin(self):
        user_pin = input("Create a pin: ")
        re_pin = input("Re-Enter your pin: ")
        if user_pin == re_pin:
            self.pin = user_pin
            print("Pin created successfully")
        else:
            print("Check your pin and try again")
        user_balance = int(input("Enter your balance: "))
        self.__balance = user_balance
        # self.menu()



    def check_balance(self):
        pin = input("Enter your pin: ")
        if pin == self.pin:
            print(f"Your balance is: {self.__balance}")
        else:
            print("Check your pin and try again!")
        #self.menu()

    def update_pin(self):
        old_pin = input("Enter your old pin: ")
        if old_pin == self.pin:
            new_pin = input("Enter your new pin: ")
            self.pin = new_pin
            print("Pin changed Successfully!")
        else:
            print("Check Your old pin and try again!")
        #self.menu()

    def withdraw(self):
        pin = input("Enter your pin: ")
        if pin == self.pin:
            withdraw_amount = int(input("Enter amount to withdraw: "))
            if self.__balance >= withdraw_amount:
                self.__balance -= withdraw_amount
                print(f"Available amount is: {self.__balance}")
            else:
                print("Insufficient balance!")
        else:
            print("check your pin and try again!")
        #self.menu()

    def deposit(self):
        pin = input("Enter your pin: ")
        if pin == self.pin:
            deposit = int(input("Enter anount to be deposited: "))
            if deposit > 0:
                self.__balance += deposit
                print(f"Available balance is {self.__balance}")
            else:
                print("Deposit must be greater then -> 0! ")
        else:
            print("Check your pin and try again!")

```
