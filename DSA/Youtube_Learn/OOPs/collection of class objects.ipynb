## Collection of class objects in OOPS


```python
class Person:
    def __init__(self, name, country):
        self.name = name
        self.country = country
```

```python
p1 = Person("Manoj", "Ukraine")
p2 = Person("Akshaya", "Uk")
p3 = Person("jyoti", "India")
```

```python
l = [p1,p2,p3]
```

```python
print(l)
```

```python
for i in l:
    print(i.name, i.country)
```
