```python

class Solution:
    def sortedArray(self, nums):
        n = len(nums)

        for i in range(n):
            isswap = False
            for j in range(n-i-1):
                if nums[j] > nums[j+1]:
                    # Swap
                    temp = nums[j]
                    nums[j] = nums[j + 1]
                    nums[j+1] = temp
                    isswap = True

            if not isswap:
                break

        return nums


obj = Solution()
nums = list(map(int, input("Enter List of numbers: ").split()))
print(obj.sortedArray(nums))

```
